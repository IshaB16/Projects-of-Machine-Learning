```python
X = df[
[
    "Type","Air temperature [K]","Process temperature [K]","Rotational speed [rpm]","Torque [Nm]","Tool wear [min]"
]
]
```

```python
y = df["Machine failure"]
```

```python
print("Input shape:", X.shape)
print("Output shape:",y.shape)
```

```python
categorical_features = ["Type"]

numerical_features = [
    "Air temperature [K]","Process temperature [K]","Rotational speed [rpm]","Torque [Nm]","Tool wear [min]"
]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)

print("Training samples:",X_train.shape[0])
print("Testing samples:",X_test.shape[0])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),numerical_features),
        ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_features)
    ]
)
```

```python
linear_svm = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier", SVC(kernel="linear"))]
)
linear_svm.fit(X_train,y_train)
```

```python
y_pred_linear = linear_svm.predict(X_test)
```

```python
accuracy_linear = accuracy_score(y_test, y_pred_linear)
precision_linear = precision_score(y_test, y_pred_linear)
recall_linear = recall_score(y_test, y_pred_linear)
f1_linear = f1_score(y_test, y_pred_linear)

print("Accuracy:", accuracy_linear)
print("Precision:", precision_linear)
print("Recall:", recall_linear)
print("F1 Score:", f1_linear)
```

```python
print(
    classification_report(y_test,y_pred_linear,target_names=["No Failure","Failure"],zero_division=0)
)
```

```python
cm_linear = confusion_matrix(y_test,y_pred_linear)

plt.figure(figsize=(6,5))

sns.heatmap(cm_linear,annot=True,fmt="d",cmap="Blues",xticklabels=["No Failure","Failure"],yticklabels=["No Failure","Failure"])

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Linear SVM")

plt.show()
```

```python
rbf_svm = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier",SVC(kernel="rbf"))]
)
rbf_svm.fit(X_train,y_train)
```

```python
y_pred_rbf = rbf_svm.predict(X_test)
```

```python
accuracy_rbf = accuracy_score(y_test, y_pred_rbf)
precision_rbf = precision_score(y_test, y_pred_rbf)
recall_rbf = recall_score(y_test, y_pred_rbf)
f1_rbf = f1_score(y_test, y_pred_rbf)

print("RBF SVM Results")
print("Accuracy:", accuracy_rbf)
print("Precision:", precision_rbf)
print("Recall:", recall_rbf)
print("F1 Score:", f1_rbf)

```

```python
print(classification_report(y_test,y_pred_rbf,target_names=["No Failure", "Failure"],zero_division=0))
```

```python
cm_rbf= confusion_matrix(y_test,y_pred_rbf)

plt.figure(figsize=(6,5))

sns.heatmap(cm_rbf,annot=True,fmt="d",cmap="coolwarm",xticklabels=["No Failure","Failure"],yticklabels=["No Failure","Failure"])

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - RBF SVM")

plt.show()
```

```python
comparison = pd.DataFrame({
    "Kernel":["Linear","RBF"],
    "Accuracy": [accuracy_linear,accuracy_rbf],
    "Precision": [precision_linear,precision_rbf],
    "Recall": [recall_linear,recall_rbf],
    "F1 Score": [f1_linear,f1_rbf]
})
```

```python
comparison_plot = comparison.set_index("Kernel")

comparison_plot.plot(
    kind="bar",figsize=(10,6)
)

plt.title("Comparison of SVM Kernels")
plt.xlabel("SVM Kernel")
plt.ylabel("Score")
plt.ylim(0,1)

plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1),loc="upper left")

plt.tight_layout()
plt.show()
```

```python
pipeline = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier",SVC())]
)
```

```python
param_grid = [{"classifier_kernel":["linear"],"classifier_C":[0.1,1,10,100]},
             {
                 "classifier_kernel": ["rbf"],"classifier_C":[0.1,1,10,100],"classifier_gamma": ["scale",0.01,0.1,1]
             }]
```

```python
grid_search = GridSearchCV(estimator=pipeline,param_grid=param_grid,cv=5,scoring="f1",n_jobs=-1,verbose=1)
```

```python
grid_search.fit(X_train,y_train)

print("GridSearchCV completed.")
```
